```python
#Generador congruencial
#Programa que imprime numeros aleatorios por medio de generadores congruenciales
```

```python
def generar(a,b,m,x0):
  X = x0
  X = (a * X + b) % m
  return(X)


a=8
b=16
m=100
x0=15
n=10
muestra = []
for i in range(n):
  nuevo = generar(a,b,m,x0)
  muestra.append(nuevo)
  x0=nuevo
print(muestra)
```
