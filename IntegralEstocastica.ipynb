<h1> Integral Estocastica </h1>

Codigo que resuelve el problema de simulacion con $t=1$ de la actividad a entregar



```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def browniano(T,n):
  dt = T / n
  B = np.zeros(n+1)
  for i in range(n):
    Z = np.random.normal(0,1)
    B[i+1] = B[i] + np.sqrt(dt) * Z
  return B
```

```python
def integral(a,b,n):
    inte = 0
    dt = (b-a)/n
    B = browniano(b-a,n)

    for i in range(n):
        inte += dt * (B[i]**2)

    return inte

I = integral(0,1,1000)
print("Valor integral:", I)
```

```python
N = 10000

lista = []

for i in range(N):
    I = integral(0,1,1000)
    lista.append(I)

media = np.mean(lista)
varianza = np.var(lista)

print("Esperanza simulada:", media)
print("Esperanza teórica:", 1/2)

print("Varianza simulada:", varianza)
print("Varianza teórica:", 1/3)
```
